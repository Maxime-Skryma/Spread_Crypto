```python
r=0.05 #Taux d'intérêt
S0=100 #Prix de départ de l'actif

u=1.5
d=0.5

if d>=1+r or u<= 1+r:
    raise ValueError("Ne respecte pas les conditions d'absence d'arbitrage")

K=100 #strike
#
S1_u=u*S0
S1_d=d*S0

C1_u=max(S1_u - K,0)
C1_d=max(S1_d - K,0)
```

```python
q=(1+r-d)/(u-d)

prix_1=(1/(1+r))*(q*C1_u + (1-q)*C1_d)

delta=(C1_u-C1_d)/(S1_u-S1_d) #Variation de l'option par  rapport au sous jacent
B=(C1_u-delta*S1_u)/(1+r) #montant dans l'actif sans risque


prix_2= delta*S0 + B #Le prix du call est le prix du portfeuille de réplication : le portfeuille tel que dans les deux issus on reçoit ce que nous aurait donné l'option.

print(prix_1,prix_2)
```

```python
def Call_Binomial_1_Period(S0,K,r,u,d):


    if d>=1+r or u<= 1+r:
        raise ValueError("Ne respecte pas les conditions d'absence d'arbitrage")
    S1_u=u*S0
    S1_d=d*S0

    C1_u=max(S1_u - K,0)
    C1_d=max(S1_d - K,0)
    
    q=(1+r-d)/(u-d) #Calcul de la proba risque-neutre

    prix_call=(1/(1+r))*(q*C1_u + (1-q)*C1_d)

    return("Le prix du call en absence d'arbitrage est ", prix_call)
```

```python
#Paramètres 
S0=100 #Spot à la période 0
N=4 #Nombre de périodes
K=120 #Strike

r=0.03
u=1.5
d=0.5


def Call_Binomial_N_Period_retro(N,S0,K,r,u,d):


    if d>=1+r or u<= 1+r:
        raise ValueError("Ne respecte pas les conditions d'absence d'arbitrage")

    SN=[ S0*(u**k)*(d**(N-k)) for k in range(0,N+1)] #Vecteur des valeurs possibles du Spot à la dernière période
    CN=[ max(S0*(u**k)*(d**(N-k))-K,0) for k in range(0,N+1)] #Vecteurs des payoffs à la dernières période
    
    q=(1+r-d)/(u-d) #Calcul de la proba risque-neutre

    for t in range(N,0,-1):
        C_N_minus_1=[0]*t
        for k in range(0,t):
            C_N_minus_1[k]=(1/(1+r))*(q*CN[k+1] + (1-q)*CN[k])
        CN=C_N_minus_1
    C0=CN[0]
    return C0
```

```python
def Call_Binomial_N_Period_closed(N,S0,K,r,u,d):


    if d>=1+r or u<= 1+r:
        raise ValueError("Ne respecte pas les conditions d'absence d'arbitrage")

    SN=[ S0*(u**k)*(d**(N-k)) for k in range(0,N+1)] #Vecteur des valeurs possibles du Spot à la dernière période
    CN=[ max(S0*(u**k)*(d**(N-k))-K,0) for k in range(0,N+1)] #Vecteurs des payoffs à la dernières période
    
    q=(1+r-d)/(u-d) #Calcul de la proba risque-neutre

    C0=0 

    for k in range(0,N+1):
        C0+=math.comb(N,k)*(q**k)*((1-q)**(N-k))*CN[k]

    C0=C0/((1+r)**N)
    return C0
```

```python
def American_Put_Binomial_N_Period(N,S0,K,r,u,d):


    if d>=1+r or u<= 1+r:
        raise ValueError("Ne respecte pas les conditions d'absence d'arbitrage")

    SN=[ S0*(u**k)*(d**(N-k)) for k in range(0,N+1)] #Vecteur des valeurs possibles du Spot à la dernière période
    PN=[ max(K-S0*(u**k)*(d**(N-k)),0) for k in range(0,N+1)] #Vecteurs des payoffs à la dernières période
    
    q=(1+r-d)/(u-d) #Calcul de la proba risque-neutre

    P0=0
    
    for t in range(N,0,-1):
        P_N_minus_1=[0]*t
        SN=[ S0*(u**k)*(d**(t-1-k)) for k in range(0,t)]
        for k in range(0,t):
            P_N_minus_1[k]=max((1/(1+r))*(q*PN[k+1] + (1-q)*PN[k]),K-SN[k])

        PN=P_N_minus_1
    P0=PN[0]
    return P0
```

```python
def Simulation_Black_Scholes(T,sigma,S0,K,r,nbr):
    moyenne=(r-(sigma**2)/2)*T
    variance_sqrt=np.sqrt(T*sigma**2)
    simulation=rng.normal(moyenne,variance_sqrt,nbr)

    exp_simulation=np.exp(simulation)
    ST_simulation=S0*exp_simulation
    C_simulation=np.maximum(ST_simulation-K,0)

    borne=1.96*(np.sqrt(np.var(C_simulation,ddof=1)/nbr))*np.exp(-r*T)

    plt.hist(ST_simulation,bins='auto',density=True)
    plt.title('Distribution de ST')
    plt.xlabel('Valeur')
    plt.ylabel('Densité')
    plt.show()
    return [{np.mean(C_simulation)*np.exp(-r*T) - borne},{np.mean(C_simulation)*np.exp(-r*T) + borne}]
```

```python
# T,sigma,S0,K,r
# d1
K=120

def Delta(T,sigma,S0,K,r):
    d1=(np.log(S0/K)+(r+(sigma**2)/2)*T)/(sigma*np.sqrt(T))
    return norm.cdf(d1)

def Gamma(T,sigma,S0,K,r):
    d1=(np.log(S0/K)+(r+(sigma**2)/2)*T)/(sigma*np.sqrt(T))
    return norm.pdf(d1)/(S0*sigma*np.sqrt(T))

def Vega(T,sigma,S0,K,r):
    d1=(np.log(S0/K)+(r+(sigma**2)/2)*T)/(sigma*np.sqrt(T))
    return norm.pdf(d1)*S0*np.sqrt(T)

def Theta(T,sigma,S0,K,r):

    d1=(np.log(S0/K)+(r+(sigma**2)/2)*T)/(sigma*np.sqrt(T))
    d2=d1-sigma*np.sqrt(T)

    terme1=-norm.pdf(d1)*S0*sigma/(2*np.sqrt(T))
    terme2=-r*K*np.exp(-r*T)*norm.cdf(d2)
    return terme1+terme2

def Greeks_plots(T,sigma,K,r): #En fonction de S0
    S0=np.linspace(80,160,100)
    delta=Delta(T,sigma,S0,K,r)
    gamma=Gamma(T,sigma,S0,K,r)
    vega=Vega(T,sigma,S0,K,r)
    theta=Theta(T,sigma,S0,K,r)

    greeks={
        "delta":delta,
        "gamma":gamma,
        "vega":vega,
        "theta":theta
    }
    for names,g in greeks.items():
        plt.plot(S0,g)
        plt.title(f"Evolution du {names} en fonction de S0")
        plt.xlabel('S0')
        plt.ylabel(names)
        plt.show()
```

```python
Greeks_plots(T,sigma,K,r)
```

```python
S0=100
def Greeks_plots_T(S0,sigma,K,r): #En fonction de S0
    T=np.linspace(0.1,8,100)

    delta=[0]*3
    gamma=[0]*3
    vega=[0]*3
    theta=[0]*3
    a=0
    for strike in [80,100,120]:

        delta[a]=Delta(T,sigma,S0,strike,r)
        gamma[a]=Gamma(T,sigma,S0,strike,r)
        vega[a]=Vega(T,sigma,S0,strike,r)
        theta[a]=Theta(T,sigma,S0,strike,r)
        a+=1

    greeks={
        "delta":delta,
        "gamma":gamma,
        "vega":vega,
        "theta":theta
    }
    for names,g in greeks.items():
        plt.plot(T,g[0],label="K=80")
        plt.plot(T,g[1],label="K=100")
        plt.plot(T,g[2],label="K=120")
        plt.legend()
        plt.title(f"Evolution du {names} en fonction de T")
        plt.xlabel('T')
        plt.ylabel(names)
        plt.show()
```
